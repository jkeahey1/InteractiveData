
### What does the "write/render/evaluate" loop provide for creators of interactive data visualizations?
The write/render/evaluate loop allows for refinement of a desigN, By writing code, then rendering it, we can see what assess what could be improved upon and what is already working. When it comes to interactive data one of the potential refinement areass is if our data viz is easily understood by anyone who is not familar with our data set. 

### Why are dynamic, interactive data visualizations useful when communicating data?
Interactive data visulatizations are useful when communicating data as it can break down large amounts of data. Looking at data sets over years or with multiple variables can be extremely overwhemling but the interactivity allows users to dive deeper and get more details on a data set. This aligns with the, “'Visual Information-Seeking Mantra”: overview first, zoom and filter, then details-on-demand." Static images do not allow users to filter and get details on demand.
